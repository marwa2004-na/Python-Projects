```python
print("Hello,if you want to calculate a matrix of 2X2 order fill the following:") 
print("__________________________________________________________________________") 
 
matrix1=[[],[]] 
 
for i in range(4): 
    a=int(input("Enter matrix1 values:")) 
    if i<2: 
        matrix1[0].append(a) 
    else: 
        matrix1[1].append(a) 
 
print(matrix1[0])  
print(matrix1[1]) 
 
print("_______________________________") 
 
matrix2=[[],[]] 
 
for i in range(4): 
    a=int(input("Enter matrix2 values:")) 
    if i<2: 
        matrix2[0].append(a) 
    else: 
        matrix2[1].append(a) 
 
print(matrix2[0]) 
 
print(matrix2[1]) 
result = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] += matrix1[i][k] * matrix2[k][j]
        
print("Result of matrix multiplication:")
print(result[0])
print(result[1])
```
