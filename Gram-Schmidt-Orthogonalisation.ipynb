```python
#get array,vectors library 
import numpy as np

#create function to do gram-schmidt orthogonalisation
def gram_schmidt(vectors):
    vectors = [v.astype(float) for v in vectors]  # convert to float
    #create lists to store values 
    orthogonal = []
    orthonormal = []
    
    #a loop to traverse each value and get orthogonal basis set
    for v in vectors:
        #copy the origional set to do calculation and leave origional intact
        orth_set = v.copy()
        for u in orthogonal:
            #orthogonal projection Yn=Xn-(<Xn,Yn>/<Yn,Yn>)*Yn
            orth_set = orth_set - np.dot(v, u) / np.dot(u, u) * u
        orthogonal.append(orth_set)
        
        #get orthognormal basis set 
        # 1e-10 to avoid divide by zero (MAth Error)
        if np.linalg.norm(orth_set) > 1e-10:
            #calculate orthonormal Un=Yn/||Yn||
            orthonormal.append(orth_set / np.linalg.norm(orth_set))
    #return function
    return orthogonal, orthonormal


# ----------User Input----------

#ask user of vector dimension and number of vector 
#loop to keep asking user to enter correct dimension value
while True:
        dim = int(input("Enter vector dimension (2 or 3): ").strip())
        if dim in (2, 3):
            break
        else:
            print("Invalid dimension. Please enter 2 or 3.\n")
    
n = int(input("Enter number of vectors: "))

#loop to get all values entered by user and turn to float value and match the dimension given
vectors = []
valid_input = True     # flag to track correctness
for i in range(n):
    vec = input(f"Enter vector {i+1} components separated by space: ").split()
    vec = np.array([float(x) for x in vec])
    
    if len(vec) == dim:
            vectors.append(vec)
    else:
        print(f"Vector dimension must be {dim}. Please re-enter.\n")
        valid_input = False
        break

# ----------Gram-Schmidt Output----------

#call funtion for both sets and print output if everything is correct
if valid_input:
    orthogonal, orthonormal = gram_schmidt(vectors)

    print("\nOrthogonal basis:")
    for i in range(len(orthogonal)):
        print(f"y{i+1} =", orthogonal[i])

    print("\nOrthonormal basis:")
    for i in range(len(orthonormal)):
        print(f"u{i+1} =", orthonormal[i])
else:
    print("No output due to incorrect vector dimension.")
```
