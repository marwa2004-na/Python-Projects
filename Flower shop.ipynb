```python
print("\t\t\t\t\t\tThe Flower Shop")
print("\t\t\t\t\t\t*****************")
def welcome():
    print("\t\tWelcome to our online-based shop for flowers, Hope you enjoy shopping with us.") # Introduction to the shop
    print("\t\t______________________________________________________________________________\n")
welcome()
h = input("Is it your first visit to our shop?\n ")# Welcoming visitors for both the first visitor and those who came back
hi = lambda: "Welcome to our shop\n" if h.lower() == "yes" else "Welcome back\n"
print(hi())
print("___________________________")
 
 
 #user details:
print("\n")
print("Enter your details: ")
name= input("Enter your full name: ")
email=input("Enter Email Address: ")
 
 
 #selection of flowers
print("_______________________________________________________________________________")
flowers = ['1.Daises-5$ ', '2.Sunflowers-3$ ', '4.Lavender-6$ ','5.Tulips-9$'] # Flower types that the customer can order
flowers.pop()
flowers.insert(2,'3.Liles-10$ ')
flowers.append('5.Roses-15$')
 
print("\nThe flowers we have are: ")
for i in flowers:
    print(i)
 
 #calculating the amount
 
def total(): # Function to calculate the total cost of the order
    total_cost = 0
    for i in range(a):
        s = int(input("Choose the flower type by number: "))
        if s == 1:
            total_cost += 5
        elif s == 2:
            total_cost += 3
        elif s == 3:
            total_cost += 10
        elif s == 4:
            total_cost += 6
        elif s == 5:
            total_cost += 15
        else:
            print("Invalid Flower type. Please choose again")
            continue
    return total_cost
 
a = int(input("\nEnter the number of types you want to buy: "))
print("\nThe total cost is:", total(), "$")
print("______________________________________________________________________")
 
#customer location details
print("\n")
print("Enter your location details please")
print("*"*50)
address1=input("Enter Address Line 1: ")
address2=input("Enter Address Line 2(optional): ")
city=input("Enter City: ")
state=input("Enter State/Region: ")
postalcode=int(input("Enter Postal Code: "))
phoneno=int(input("Enter Phone number: "))
special=input("Enter any special instructions(optional): ")
 
delivery_address={
    'Address Line 1':address1,
    'Address Line 2':address2 if address2 else None,
    'City':city,
    'State/Region':state,
    'Postal Code':postalcode,
    'Phone number':phoneno,
    'Special Instruction':special if special else None
}
delivery_address.popitem()

 
 
print("____________________________________________________________________________________")
print("\n")
#updating customer details
def customer(email,address1,city,state,postalcode,phoneno):
    print("Customer Details-")
    do=input("Do you want to update your details:")
    if do.lower()=="yes":
        print("1.Email\n2.Address Line 1\n3.City\n4.State\n5.Postalcode\n6.Phone number")
        which=int(input("Which one do you want to change (choose by number):"))
        if which==1:
            up=input("Update your Email address:")
            delivery_address["Email"]=up
            return delivery_address
        elif which==2:
            up=input("Update your Address Line 1:")
            delivery_address["Address Line 1"]=up
            return delivery_address
        elif which==3:
            up=input("Update your City:")
            delivery_address["City"]=up
            return delivery_address
        elif which==4:
            up=input("Update your State:")
            delivery_address["State/Region"]=up
            return delivery_address
        elif which==5:
            up=input("Update your Postal code:")
            delivery_address["Postal code"]=up
            return delivery_address
        elif which==6:
            up=input("Update your Phone number:")
            delivery_address["Phone number"]=up
            return delivery_address
    else:
        return delivery_address
customer(email,address1,city,state,postalcode,phoneno)
delivery_address.update({"Email":email})
print("____________________________________________________________________________________")
print("\n")
print(f"Thank you,{name}.Your details are listed below: ")
for key, value in delivery_address.items():
    print(f"{key}: {value}")
 
print("\n")
 
print("____________________________________________________________________________________")
 
 #function for the customer to choose the payment method
print("\n")
def options():
 
    print(f"Dear customer {name},how do you want to pay(cash/bank):")
    print("1. Cash on delivery")
    print("2. Via bank transfer")
    payment=input("Enter your preferred method: ").lower()
 
    if payment=="cash":
        print("You've chosen to pay with cash. Please keep the cash ready for the delivery person.")
    elif payment=="bank":
        bank_name=input("Enter your bank name: ")
        if bank_name:
            bank_accountno=input(f"Enter your account number for {bank_name}")
            print(f"Thank you for providing your bank details for {bank_name}.")
        else:
            print("Bank name cannot be empty. Please provide a valid bank name.")
    else:
        print("Invalid payment method. Please choose either 'cash' or 'bank'.")

options()
print("\n")
 
print("____________________________________________________________________________________")
 
 
 #confirmation of order:
print("\n")
 
def confirm(name,order):
    if order=="yes":
        print(f"Thank you, {name}! Your order has been confirmed.")
        print("You can track your order through our app.")
    else:
        print(f"Thank you, {name}.Your order has been canceled. We hope to see you again soon!")
 
 
order=input("Do you want to confirm this order? ").lower()
confirm(name,order)
 
print("\n")
```
