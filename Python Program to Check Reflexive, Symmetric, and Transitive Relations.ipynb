```python
# Input set A
A = set(map(int, input("Enter elements of set A (space separated): ").split()))

# Input relation R
n = int(input("Enter number of pairs in relation: "))
R = set()

print("Enter pairs like: a b")
for i in range(n):
    a, b = map(int, input().split())
    R.add((a, b))

# Reflexive
def is_reflexive(A, R):
    for a in A:
        if (a, a) not in R:
            return False
    return True

# Symmetric
def is_symmetric(R):
    for (a, b) in R:
        if (b, a) not in R:
            return False
    return True

# Transitive
def is_transitive(R):
    for (a, b) in R:
        for (c, d) in R:
            if b == c and (a, d) not in R:
                return False
    return True

# Output
print('A:',A)
print('R:',R)
print("Reflexive:", is_reflexive(A, R))
print("Symmetric:", is_symmetric(R))
print("Transitive:", is_transitive(R))
```
